## <font color="#3366FF">LangChain Document Loaders</font>

### Example-1 Load PDF File With PyPDFLoader

```python
import os
from langchain_classic.document_loaders import PyPDFLoader

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "sample.pdf")

## Initialize Document Loader & Load Document
loader = PyPDFLoader(file_path)
docs = loader.load()

## Print output
print(f"Loaded {len(docs)} document(s)")
print("Metadata:", docs[0].metadata)
```

#### What’s Happening Here?

```python
## Document Loader Returns List of Document
print(type(docs))
print(type(docs[0]))
```

```python
print(docs)
```

#### View Content

```python
print(docs[0].page_content)
```

#### Async Load 

```python
import os
from langchain_classic.document_loaders import TextLoader

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "sample.txt")

## Initialize Document Loader & Load Document
loader = TextLoader(file_path)
docs = await loader.aload()

## Print output
print(f"Loaded {len(docs)} document(s)")
print("Metadata:", docs[0].metadata)
```

#### Lazy Load

```python
import os
from langchain_classic.document_loaders import PyPDFLoader

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "sample.pdf")

## Initialize Document Loader
loader = PyPDFLoader(file_path)

## Lazy load document
docs_iterator = loader.lazy_load()

## Loop through documents one-by-one
for doc in docs_iterator:
    print("Document content:", doc.page_content)
```

## <font color="#3366FF">JSON Loader</font>

#### Content Of JSON File

### Load Entire JSON As Single LangChain Document

```python
import os
from langchain_classic.document_loaders import JSONLoader

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "sample.json")

## Initialize Document Loader & Load Document
loader = JSONLoader(
    file_path=file_path,
    jq_schema= ".",
    text_content=False
)
docs = loader.load()

## Print output
print(f"Loaded {len(docs)} document(s)")
print("Metadata:", docs[0].metadata)
print("Content:")
print(docs[0].page_content)
```

### Load Each JSON Array Record As LangChain Document

```python
import os
from langchain_classic.document_loaders import JSONLoader

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "sample.json")

## Initialize Document Loader & Load Document
loader = JSONLoader(
    file_path=file_path,
    jq_schema= ".[]",
    text_content=False
)
docs = loader.load()

## Print output
print(f"Loaded {len(docs)} document(s)")
print("Metadata:", docs[0].metadata)
print("Content:")
print(docs[0].page_content)
```

## <font color="#3366FF">Directory Loader</font>

```python
import os
from langchain_classic.document_loaders import DirectoryLoader, TextLoader

## Get the current working directory
cwd = os.getcwd()

## Set your documents folder path
folder_path = os.path.join(cwd, "documents")

## Initialize the DirectoryLoader to load only .txt files using TextLoader
loader = DirectoryLoader(
    path=folder_path,
    glob="**/*.txt",  # Recursive pattern for .txt files
    loader_cls=TextLoader,  # Use TextLoader for each file
    show_progress=True      # Optional: shows a progress bar
)

## Load documents
docs = loader.load()

## Print output
print(f"Loaded {len(docs)} document(s)")

# Iterate and print metadata
for idx, doc in enumerate(docs, start=1):
    print(f"Document {idx}:")
    print("Metadata:", doc.metadata)
    print("-" * 60)
```
