# <font color="#800000">Motivation For Type Hinting In Python</font>

```python
## Assign a string to the variable
str = "Good Morning"
print("type of str is ", type(str))

## ... <later in code>....

## Assign an integer to the variable
str = 25
print("type of str is ", type(str))
```

## <font color="#3366FF">Python Type Hinting</font>

```python
## Built-in Type Hints in Python
age: int = 24

currency: float = 100.25

name: str = "Bob"

is_valid: bool = True
```

```python
## Define Function with Type Hints
def add(a: int, b: int) -> int:
    return a + b
```

### Important: Type Hints are Not Enforced

```python
## Annotate age variable as int
age: int

## Assign a string to age
age = "Good Morning"

## Python treats age as a string type
print("type of age is ", type(age))
```

# <font color="#800000">Python Typing Library</font>

## <font color="#3366FF">TypedDict</font>

```python
from typing import TypedDict

## Define a custom class Person using TypedDict
class Person(TypedDict):
    name: str
    age: int

## Define variable x as a type Person
x: Person = {"name": "Alice", "age": 30}
```

## <font color="#3366FF">Annotated</font>

```python
from typing import TypedDict, Annotated

## Define a custom class Person using TypedDict
class Person(TypedDict):
    name: Annotated[str, "The full name"]
    age: Annotated[int, "Must be a positive integer"]

## Declare variable x as a type Person
x: Person = {"name": "Alice", "age": 30}
```

## <font color="#3366FF">Optional</font>

```python
from typing import TypedDict, Optional

## Define a custom class UserProfile using TypedDict
class UserProfile(TypedDict):
    name: str
    age: int
    phone: Optional[str]  # Phone can be a string or None

## Declare variables user1 & user2 as a type of UserProfile
user1: UserProfile = {"name": "Alice", "age": 30, "phone": "555-1234"} 
user2: UserProfile = {"name": "Alice", "age": 30, "phone": None}
```

## <font color="#3366FF">List</font>

```python
from typing import TypedDict, List

## Define a custom class UserProfile using TypedDict
class UserProfile(TypedDict):
    name: str
    age: int
    interest: List[str]  # A list of strings

## Declare a variable user as a type of UserProfile
user: UserProfile = {"name": "Alice", "age": 30, "interest": ["Travel","Yoga","Data Science"]}
```

## <font color="#3366FF">Dict</font>

```python
from typing import TypedDict, Dict

## Define a custom class UserProfile using TypedDict
class UserProfile(TypedDict):
    name: str
    age: int
    marks: Dict[str, int]  # Dictionary with string keys and integer values

## Declare a variable user as a type of UserProfile
user: UserProfile = {"name": "Alice", "age": 30, "marks": {"Math": 80, "English": 72, "Science": 89}}
```

## <font color="#3366FF">Literal</font>

```python
from typing import TypedDict, Literal

## Define a custom class UserProfile using TypedDict
class UserProfile(TypedDict):
    name: str
    age: int
    grade: Literal["A", "B", "C"] # Possible values of Grade

## Declare a variable user as a type of UserProfile
user: UserProfile = {"name": "Alice", "age": 30, "grade": "B"}
```

# <font color="#800000">Python Pydantic Library</font>

## <font color="#3366FF">BaseModel</font>

```python
from pydantic import BaseModel

## Define a custom class UserProfile using BaseModel
class UserProfile(BaseModel):
    name: str
    age: int

## Declare a variable user as a type of UserProfile
user = UserProfile(name="Phillips", age=46)  

## Print Output as Dictionary
print(user.model_dump())
```

```python
## This will throw data validation error
user = UserProfile(name="Phillips", age="Fourty Six")
```

## <font color="#3366FF">Compatibility with Typing Library</font>

```python
from pydantic import BaseModel
from typing import Optional, List, Dict, Literal

## Define a custom class UserProfile using BaseModel
class UserProfile(BaseModel):
    name: str
    age: int
    phone: Optional[str] 
    interest: List[str]
    marks: Dict[str, int] 
    grade: Literal["A", "B", "C"]
    

## Declare a variable user as a type of UserProfile
user = UserProfile(name="Phillips", age=46, phone=None, interest=["Travel","Yoga","Data Science"], 
                   marks = {"Math": 80, "English": 72, "Science": 89}, grade="C")  

## Print Output as Dictionary
print(user.dict())
```

### Data Validation Example

```python
## This will throw data validation error
user = UserProfile(name="Phillips", age=46, phone=None, interest=["Travel","Yoga","Data Science"], 
                   marks = {"Math": 80, "English": 72, "Science": 89}, grade="D")
```

## <font color="#3366FF">Field</font>

```python
from pydantic import BaseModel, Field

## Define a custom class UserProfile using BaseModel
class UserProfile(BaseModel):
    name: str = Field(..., description="Full name of the person")
    age: int = Field(default=None, ge=18, description="Age must be 18 or older")

## Declare a variable user as a type of UserProfile
user = UserProfile(name="Daisy")

## Print Output as Dictionary
print(user.model_dump())
```

#### Alternative Style

```python
from pydantic import BaseModel, Field
from typing import Annotated

## Define a custom class UserProfile using BaseModel
class UserProfile(BaseModel):
    name: Annotated[str, Field(..., description="Full name of the person")]
    age: Annotated[int, Field(default=None, ge=18, description="Age must be 18 or older")]

## Declare a variable user as a type of UserProfile
user = UserProfile(name="Daisy")

## Print Output as Dictionary
print(user.model_dump())
```

### Data Validation Example

```python
## This will throw data validation error
user = UserProfile(name="Daisy", age = 15)
```
