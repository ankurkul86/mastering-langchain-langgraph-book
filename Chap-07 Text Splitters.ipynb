### Create Document Method

```python
from langchain_text_splitters import CharacterTextSplitter

## Example text
text = """Artificial intelligence (AI) is rapidly transforming the world, enabling machines to perform tasks that traditionally require human intelligence. From voice assistants to personalized recommendations, AI is becoming increasingly integrated into our daily lives. 

AI systems use techniques like machine learning and deep learning to solve problems and make decisions, often exceeding human capabilities in speed and accuracy. 
"""

## Initialize the Text Splitter
splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20, separator='')

## Split the text
chunks = splitter.create_documents([text], metadatas=[{"author": "Ankur", "purpose":"example"}])

## Print document chunks one by one
for i, doc in enumerate(chunks):
    print(f"Document {i+1}")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}\n")
    print("---")
```

### Split Document Method

```python
import os
from langchain_classic.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "example.txt")

## Initialize Document Loader & Load Document
loader = TextLoader(file_path=file_path)
docs = loader.load()

## Initialize the Text Splitter
splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20, separator=',')

## Split the document
chunks = splitter.split_documents(docs)

## Print document chunks one by one
for i, doc in enumerate(chunks):
    print(f"Document Chunk {i+1}")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}\n")
    print("---")
```

### Load And Split Method

```python
import os
from langchain_classic.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "example.txt")

## Initialize Document Loader
loader = TextLoader(file_path=file_path)

## Initialize the Text Splitter
splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20, separator=',')

## Split the document
chunks = loader.load_and_split(text_splitter=splitter)

## Print document chunks one by one
for i, doc in enumerate(chunks):
    print(f"Document Chunk {i+1}")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}\n")
    print("---")
```

### Using Token With CharacterTextSplitter

#### Tiktoken Tokenizer

```python
from langchain_text_splitters import CharacterTextSplitter

## Example text
text = """Artificial intelligence (AI) is rapidly transforming the world, enabling machines to perform tasks that traditionally require human intelligence. From voice assistants to personalized recommendations, AI is becoming increasingly integrated into our daily lives. 

AI systems use techniques like machine learning and deep learning to solve problems and make decisions, often exceeding human capabilities in speed and accuracy. 
"""

## Initialize the Text Splitter
splitter = CharacterTextSplitter.from_tiktoken_encoder(encoding_name="cl100k_base",chunk_size=30, chunk_overlap=5, separator=',')

## Split the text
chunks = splitter.split_text(text)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
    print("---")
```

#### HuggingFace Tokenizer

```python
from langchain_text_splitters import CharacterTextSplitter
from transformers import GPT2TokenizerFast

## Initialize Tokenizer
tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")

## Example text
text = """Artificial intelligence (AI) is rapidly transforming the world, enabling machines to perform tasks that traditionally require human intelligence. From voice assistants to personalized recommendations, AI is becoming increasingly integrated into our daily lives. 

AI systems use techniques like machine learning and deep learning to solve problems and make decisions, often exceeding human capabilities in speed and accuracy. 
"""

## Initialize the Text Splitter
splitter = CharacterTextSplitter.from_huggingface_tokenizer(tokenizer,chunk_size=50, chunk_overlap=5, separator='\n')

## Split the text
chunks = splitter.split_text(text)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
    print("---")
```

## <font color="#3366FF">Markdown Header Text Splitter</font>

### Usage Of MarkdownHeaderTextSplitter

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

## Example Markdown data
md_text = """
# Product Guide

Welcome to the product guide.

## Features
- Feature A: Easy to use
- Feature B: Fast

## Notes

Please read the terms.
Contact support if needed.
"""

## Initialize the Splitter
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#", "H1"), ("##", "H2"), ("###", "H3")]
)

## Split the Markdown data
chunks = splitter.split_text(md_text)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
    print("---")
```

#### Strip Headers

```python
## <...same code here as earlier..>

## Initialize the Splitter
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#", "H1"), ("##", "H2"), ("###", "H3")], strip_headers = False
)

## Split the Markdown data
chunks = splitter.split_text(md_text)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
    print("---")
```

#### Return Each Line

```python
## <...same code here as earlier..>

## Initialize the Splitter
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#", "H1"), ("##", "H2"), ("###", "H3")], return_each_line = True
)

## Split the Markdown data
chunks = splitter.split_text(md_text)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
    print("---")
```

## <font color="#3366FF">HTML Semantic Preserving Splitter</font>

### Usage Of HTMLSemanticPreservingSplitter

```python
from langchain_text_splitters import HTMLSemanticPreservingSplitter

## Sample HTML content
html_data = """
<html>
  <body>
    <h1>Project Alpha</h1>  
    <h2>Team Members</h2>
    <ul>
      <li>Alice</li>
      <li>Bob</li>
      <li>Charlie</li>
    </ul>
    <h2>Milestones</h2>
    <table>
      <tr><th>Phase</th><th>Deadline</th></tr>
      <tr><td>Design</td><td>2025-07-01</td></tr>
      <tr><td>Development</td><td>2025-08-15</td></tr>
    </table> 
    <h2>Resources</h2>
    <p>For more info, visit <a href="https://example.com">our portal</a>.</p>
    <img src="logo.png" alt="Project Logo" />
  </body>
</html>
"""

## Initialize the semantic splitter
splitter = HTMLSemanticPreservingSplitter(
    headers_to_split_on=[("h1", "Project"), ("h2", "Section")],
    preserve_links=True,
    preserve_images=True
)

## Split the HTML content
chunks = splitter.split_text(html_data)

## Print chunks one by one
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print("Content:\n", chunk.page_content)
    print("Metadata:", chunk.metadata)
```

### Example

```python
from langchain_text_splitters import Language, RecursiveCharacterTextSplitter

## Sample Python Code
python_code = """
import random

questions = {
    "What is the capital of France?": "Paris",
    "Which planet is known as the Red Planet?": "Mars",
    "What is 5 multiplied by 6?": "30"
}

score = 0
question_list = list(questions.items())
random.shuffle(question_list)

for question, answer in question_list:
    user_answer = input(question + " ")
    if user_answer.strip().lower() == answer.lower():
        print("Correct!")
        score += 1
    else:
        print(f"Wrong! The correct answer was {answer}.")

print(f"Your final score: {score}/{len(questions)}")

"""

## Initialize RecursiveCharacterTextSplitter to Split Python Language Code
python_splitter = RecursiveCharacterTextSplitter.from_language(language=Language.PYTHON, chunk_size=250, chunk_overlap=0)

## Split Python Code
code_chunks = python_splitter.split_text(python_code)

## Print chunks one by one
for i, chunk in enumerate(code_chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print("Content:\n", chunk)
```
