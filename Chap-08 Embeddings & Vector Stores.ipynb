### Create Chroma Vector DB

```python
import os
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Define a collection name
collection_name = "test_collection"

## Create Chroma Vector Store
vectorstore = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory="chroma_db",
    collection_name=collection_name
)
```

### Create SQL Record Manager

```python
from langchain_classic.indexes import SQLRecordManager

## Define a namespace name
namespace = f"chroma/{collection_name}"

## Initialize SQLRecordManager
record_manager = SQLRecordManager(
    namespace, db_url="sqlite:///record_manager.sql"
)

## Create Schema
record_manager.create_schema()
```

### Utility Function To Print Content Of Vector Store

```python
def print_vector_store_content():
    # Get content of vector store
    vector_store_content = vectorstore.get(include=["documents", "metadatas"])

    # Check if store is empty
    if not vector_store_content["documents"] or not vector_store_content["metadatas"]:
        print("\nVector Store is Empty!\n")
        return

    print("\nContent of Vector Store:\n")

    # Iterate and print
    for doc, meta in zip(vector_store_content["documents"], vector_store_content["metadatas"]):
        print("Document Content:", doc)
        print("Metadata:", meta)
        print("-" * 50)
```

## <font color="#3366FF">Simple Data Ingestion Pipeline</font>

```python
import os
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_classic.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "example.txt")

## Load the document from a .txt file
loader = TextLoader(file_path)
documents = loader.load()

## Split the document into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,     
    chunk_overlap=5      # Overlap between chunks
)
split_docs = splitter.split_documents(documents)

## Create & persist the Chroma vector store from documents
vectorstore = Chroma.from_documents(
    documents=split_docs,            
    embedding=OpenAIEmbeddings(model="text-embedding-3-small"),      
    persist_directory="chroma_store",
    collection_name = "test_collection"
)

## Check the Content of Chroma Vector Store
results = vectorstore.get()
for doc, meta in zip(results["documents"], results["metadatas"]):
    print("Document Content:", doc)
    print("Metadata:", meta)
    print("-" * 50)
```

## <font color="#3366FF">Data Ingestion Pipeline With Indexing</font>

```python
import os
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_classic.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_classic.indexes import SQLRecordManager, index

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Get the current working directory
cwd = os.getcwd()

## Construct the full path to the file
file_path = os.path.join(cwd, "documents", "example.txt")

## Load the document from a .txt file
loader = TextLoader(file_path)
documents = loader.load()

## Split the document into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,     
    chunk_overlap=5  
)
split_docs = splitter.split_documents(documents)

## Define a collection name
collection_name = "test_collection"

## Create Chroma Vector Store
vectorstore = Chroma(
    embedding_function=OpenAIEmbeddings(model="text-embedding-3-small"),
    persist_directory="chroma_db",
    collection_name=collection_name
)

## Define a namespace name
namespace = f"chroma/{collection_name}"

## Initialize SQLRecordManager
record_manager = SQLRecordManager(
    namespace, db_url="sqlite:///record_manager.sql"
)

## Create Schema
record_manager.create_schema()

## Index Documents in Vector Store
indexing_results = index(split_docs, record_manager, vectorstore, cleanup="incremental", source_id_key="source")

## Print Indexing Result
print("Indexing Result:", indexing_results)

## Check the Content of Chroma Vector Store
results = vectorstore.get()
print("\n")
for doc, meta in zip(results["documents"], results["metadatas"]):
    print("Document Content:", doc)
    print("Metadata:", meta)
    print("-" * 50)
```

```python
import os
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_core.example_selectors import SemanticSimilarityExampleSelector
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate
from langchain_openai import OpenAIEmbeddings

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Create Few Shot Example Pool
few_shot_examples  = [
    {
        "review": "This smartwatch exceeded my expectations with its sleek design and long battery life.",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The vacuum cleaner broke down after just two uses, totally disappointed.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "The blender works fine, but it’s a bit noisier than I anticipated.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "Absolutely love the camera quality on this phone—photos are stunning even in low light!",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The laptop gets hot quickly and slows down during basic tasks.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "Received the jacket on time, fits okay, color’s a bit off from the pictures.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "These headphones deliver crisp sound and the noise cancellation is top-notch.",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The software crashes frequently and customer support is unresponsive.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "It’s just a regular notebook—nothing to rave or complain about.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "Super comfy shoes, wore them all day at the event and felt great!",
        "sentiments": "The user review is Positive",
    },
]

## Create Prompt Template that complies above Examples
few_shot_example_prompt  = PromptTemplate(
    input_variables=["review", "sentiments"],
    template="User Review: {review}\n{sentiments}",
)

## Create Semantic Similarity Example Selector
example_selector = SemanticSimilarityExampleSelector.from_examples(
    examples=few_shot_examples,
    embeddings=OpenAIEmbeddings(),
    vectorstore_cls=Chroma,
    k=4,
)

## Create Few Shot Prompt Template 
few_shot_prompt_template  = FewShotPromptTemplate(
    example_selector=example_selector,
    example_prompt=few_shot_example_prompt ,
    suffix="User Review: {user_review}",
    input_variables=["user_review"],
)

## Invoke Few Shot Prompt Template and Print
prompt = few_shot_prompt_template.invoke({"user_review": "I liked the watch, it is sleek and modern. But the dial could have been bigger"})
print(prompt.to_string())
```
