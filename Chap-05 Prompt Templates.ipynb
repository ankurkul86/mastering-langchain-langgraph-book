### Using Functions With Partial Variables

```python
from langchain_core.prompts import PromptTemplate
from datetime import datetime   

## Define function to get current date
def get_date():  
    now = datetime.now()  
    return now.strftime("%m/%d/%Y")  

## Create Prompt Template
prompt_template = PromptTemplate.from_template(template= "Give me news on {topic} for today {date}")

## Assign Partial Variables to Prompt Template
prompt_template = prompt_template.partial(date = get_date)

## Format Prompt Template
prompt_template.format(topic = "sports")
```

## <font color="#3366FF">MessagesPlaceholder</font>

### Simple Example of MessagesPlaceholder

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.messages import HumanMessage, AIMessage

## Create Prompt Template with Placeholder
prompt_template = ChatPromptTemplate([
    ("system", "You are helpful assistant"),
    MessagesPlaceholder("message_history"),
    ("user", "{user_question}")
])

## Create list of conversation history as list of Messages
conversation_history = [HumanMessage(content="How old is Earth?"),
                        AIMessage(content="The Earth is 4.5 billion years old.")]

## Invoke Prompt Template
prompt_template.invoke({"message_history": conversation_history, "user_question":"What is its diameter?"})
```

## <font color="#3366FF">FewShotPromptTemplate</font>

### Example of FewShotPromptTemplate

```python
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate

## Create Examples for Few Shots Prompt
few_shot_examples = [
    {
        "country": "Egypt",
        "answer": "Longest river of Egypt is Nile",
    },
    {
        "country": "India",
        "answer": "Longest river of India is Ganga",
    },
    {
        "country": "Bangladesh",
        "answer": "Longest river of Bangladesh is Padma",
    },
    {
        "country": "Canada",
        "answer": "Longest river of Canada is Mackenzie River",
    },
]

## Create Prompt Template that complies above Examples
few_shot_example_prompt = PromptTemplate.from_template("User Input: {country}\n{answer}")

## Create Few Shot Prompt Template 
few_shot_prompt_template = FewShotPromptTemplate(
    examples=few_shot_examples,
    example_prompt=few_shot_example_prompt,
    prefix="Give longest river of every country in input",
    suffix="User Input: {user_input}"
)


## Invoke Few Shot Prompt Template and Print
prompt = few_shot_prompt_template.invoke({"user_input": "United States"})
print(prompt.to_string())
```

### Length Based Example Selector 

#### Example of LengthBasedExampleSelector

```python
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate
from langchain_core.example_selectors import LengthBasedExampleSelector

## Create Few Shot Example Pool
few_shot_examples = [
    {
        "review": "This smartwatch exceeded my expectations with its sleek design and long battery life.",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The vacuum cleaner broke down after just two uses, totally disappointed.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "The blender works fine, but it’s a bit noisier than I anticipated.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "Absolutely love the camera quality on this phone—photos are stunning even in low light!",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The laptop gets hot quickly and slows down during basic tasks.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "Received the jacket on time, fits okay, color’s a bit off from the pictures.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "These headphones deliver crisp sound and the noise cancellation is top-notch.",
        "sentiments": "The user review is Positive",
    },
    {
        "review": "The software crashes frequently and customer support is unresponsive.",
        "sentiments": "The user review is Negative",
    },
    {
        "review": "It’s just a regular notebook—nothing to rave or complain about.",
        "sentiments": "The user review is Neutral",
    },
    {
        "review": "Super comfy shoes, wore them all day at the event and felt great!",
        "sentiments": "The user review is Positive",
    },
]

## Create Prompt Template that complies with above Examples
few_shot_example_prompt = PromptTemplate.from_template("User Review: {review}\n{sentiments}")

## Create Length Based Example Selector
length_example_selector = LengthBasedExampleSelector(
    examples=few_shot_examples,
    example_prompt=few_shot_example_prompt,
    max_length=100,
)

## Create Few Shot Prompt Template 
few_shot_prompt_template = FewShotPromptTemplate(
    example_selector=length_example_selector,
    example_prompt=few_shot_example_prompt,
    suffix="User Review: {user_review}",
)

## Invoke Few Shot Prompt Template and Print
prompt = few_shot_prompt_template.invoke({"user_review": "I liked the watch, it is sleek and modern. But the dial could have been bigger"})
print(prompt.to_string())
```

### NGram Overlap Example Selector 

#### Example of NGramOverlapExampleSelector

```python
from langchain_classic.prompts.example_selector.ngram_overlap import NGramOverlapExampleSelector
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate

## Create Few Shot Example Pool
few_shot_examples = [
    {"input": "Tell me a fact about cats", "output": "Cats sleep for 70% of their lives."},
    {"input": "Tell me a fact about elephants", "output": "Elephants can recognize themselves in a mirror."},
    {"input": "Tell me a fact about dolphins", "output": "Dolphins have names and call each other."},
    {"input": "Tell me a joke about dogs", "output": "Why did the dog sit in the shade? Because he didn’t want to be a hot dog!"},
    {"input": "Tell me a joke about fish", "output": "What do you call a fish without eyes? Fsh."},
    {"input": "Tell me a joke about parrots", "output": "Why did the parrot get a timeout? For using fowl language!"}
]

## Create Prompt Template that complies above Examples
example_prompt = PromptTemplate(
    input_variables=["input", "output"],
    template="User: {input}\nBot: {output}"
)

## Create Ngram Overlap Based Example Selector
ngram_example_selector = NGramOverlapExampleSelector(
    examples=few_shot_examples,
    example_prompt=example_prompt,
    threshold=0.5  # Lowered to allow for approximate matching
)

## Create Few Shot Prompt Template 
few_shot_prompt_template = FewShotPromptTemplate(
    example_selector=ngram_example_selector,
    example_prompt=example_prompt,
    prefix="The following are examples of user queries and bot responses:",
    suffix="User: {input}\nBot:"
)


## Invoke Few Shot Prompt Template and Print
prompt = few_shot_prompt_template.invoke({"input": "Tell me a fact on whales"})
print(prompt.to_string())
```
