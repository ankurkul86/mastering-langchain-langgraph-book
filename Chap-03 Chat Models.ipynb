## <font color="#3366FF">Non-Streaming Method</font>

### Invoke Method

```python
import os
from dotenv import load_dotenv
import json
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage, SystemMessage

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Creating Message
messages = [
    SystemMessage("You are science educator"),
    HumanMessage("Tell me a random science fact on Universe in one line."),
]

## Invoking Chat Model LLM
llm.invoke(messages)
```

#### Accessing Content, Meta Data, And Token Usage

```python
## Invoking Chat Model LLM
response = llm.invoke(messages)

print("\ncontent:")
print(response.content)

print("\nid:")
print(response.id)

print("\nresponse_metadata:")
print(json.dumps(response.response_metadata, indent=4))

print("\nusage_metadata:")
print(json.dumps(response.usage_metadata, indent=4))
```

### Async Invoke Method

```python
import os
from dotenv import load_dotenv
import json
from langchain_openai import ChatOpenAI

# Load OpenAI API key from the .env file
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

# Create OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

# Invoke Chat LLM asynchronously
response = await llm.ainvoke("Tell me a random science fact about the Universe in one line.")

# Print the outputs
print("\ncontent:")
print(response.content)

print("\nid:")
print(response.id)

print("\nresponse_metadata:")
print(json.dumps(response.response_metadata))

print("\nusage_metadata:")
print(json.dumps(response.usage_metadata))
```

### Accessing Metadata And Token Usage

```python
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import json

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Variable for aggregating streaming chunks
output = None

## Streaming Output from Chat Model
for chunk in llm.stream("Write a poem on Tom and Jerry in four lines", stream_usage=True):
    print(chunk.content, end="", flush=True)
    if output is None:
        output = chunk
    else:
        output = output + chunk

## Print Output        
print("\n\nid:")
print(output.id)

print("\nresponse_metadata:")
print(json.dumps(output.response_metadata))

print("\nusage_metadata:")
print(json.dumps(output.usage_metadata))
```

### Async Stream Method

```python
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import json

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Variable for aggregating streaming chunks
output = None

## Streaming Output from Chat Model
async for chunk in llm.astream("Write a poem on Tom and Jerry in four lines", stream_usage=True):
    print(chunk.content, end="", flush=True)
    if output is None:
        output = chunk
    else:
        output = output + chunk

## Print Output        
print("\n\nid:")
print(output.id)

print("\nresponse_metadata:")
print(json.dumps(output.response_metadata))

print("\nusage_metadata:")
print(json.dumps(output.usage_metadata))
```

## <font color="#3366FF">Batch Processing</font>

### Batch Method

```python
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Creating Batch Input with List of Prompts
batch_inputs = [
    [HumanMessage(content="Tell me a random joke")],
    [HumanMessage(content="Tell me a random science fact in one line")],
    [HumanMessage(content="Tell me a random myth in one line")],
]

## Sending Batch Request to Chat Model
responses = llm.batch(batch_inputs)
responses
```

```python
## Printing Content
for idx, response in enumerate(responses):
    print(f"Content {idx + 1}: {response.content}")
```

### Async Batch Method

```python
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Creating Batch Input with List of Prompts
batch_inputs = [
    [HumanMessage(content="Tell me a random joke")],
    [HumanMessage(content="Tell me a random science fact in one line")],
    [HumanMessage(content="Tell me a random myth in one line")],
]

## Sending async batch request
responses = await llm.abatch(batch_inputs)

## Printing Content
for idx, response in enumerate(responses):
    print(f"Response {idx + 1}: {response.content}")
```

## <font color="#3366FF">Rate Limiter</font>

```python
import os
import time
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.rate_limiters import InMemoryRateLimiter

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Defining Parameters for limiting rate
rate_limiter = InMemoryRateLimiter(requests_per_second=0.2, check_every_n_seconds=0.1, max_bucket_size=5)

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-4o-mini",temperature=1, max_tokens=512, timeout=60, max_retries=5, rate_limiter=rate_limiter, api_key=os.environ["OPENAI_API_KEY"])

## Calling LLM in a loop and recording time interval
for _ in range(3):
    start_time = time.time()
    response = llm.invoke("Tell me a random quote on life in one line")
    end_time = time.time()
    print(end_time - start_time)
    print(response.content)
```

## <font color="#3366FF">Example Of Multimodal Input</font>

### Passing Image URL As Input

#### Input Image

Source: https://st5.depositphotos.com/1007930/62661/i/450/depositphotos_626619490-stock-photo-portrait-romantic-girl-woods.jpg

![Girl in the woods with a straw hat](https://st5.depositphotos.com/1007930/62661/i/450/depositphotos_626619490-stock-photo-portrait-romantic-girl-woods.jpg)

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

## Setting Image URL
image_url = 'https://st5.depositphotos.com/1007930/62661/i/450/depositphotos_626619490-stock-photo-portrait-romantic-girl-woods.jpg'

##Creating Human Message with both text and image URL
message = HumanMessage(
    content=[
        {
            "type": "text", 
            "text": "Describe this image and list down all objects"
        },
        {
            "type": "image_url", 
            "image_url": {"url": image_url}
        },
    ],
)

## Invoking Chat Model
response = llm.invoke([message])

## Print Response
print(response.content)
```

### Passing Local Image As Input

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage
import base64

## Loading OpenAI API key from env file into an environment variable
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv('OPENAI_API_KEY')

# Path to your image
image_path = "image.png"

# Function to encode the image
with open(image_path, "rb") as image_file:
    base64_image = base64.b64encode(image_file.read()).decode("utf-8")


## Creating OpenAI Chat Model instance
llm = ChatOpenAI(model="gpt-5-mini", api_key=os.environ["OPENAI_API_KEY"])

##Creating Human Message with both text and image URL
message = HumanMessage(
    content=[
                {
                    "type": "text",
                    "text": "What season is it in the image?",
                },
                {
                    "type": "image_url",
                    "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"},
                },
    ],
)

## Invoking Chat Model
response = llm.invoke([message])

## Print Response
print(response.content)
```
